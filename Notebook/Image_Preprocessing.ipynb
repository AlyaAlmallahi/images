# Image Preprocessing Techniques
This notebook demonstrates various image preprocessing techniques using Python and OpenCV. The techniques include contrast adjustment, edge enhancement, color space conversion, noise reduction, binarization, resizing & normalization, and sharpening.

```python
# Import required libraries
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

## Load and Display the Image
We load the image `Animals.jpg` and resize it to 256x256 pixels for consistency across all techniques.

```python
# Load and resize image
 # Load and resize image
image = cv2.imread('../dataset-min/Animals.jpg')

if image is None:
    print("there is no images")
else:
    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    image = cv2.resize(image, (256, 256))

    plt.imshow(image)
    plt.title("Original Image")
    plt.axis('off')
    plt.show()

```

## Contrast Adjustment
Adjust the contrast of the image using the formula: `new_pixel = alpha * pixel + beta`.

```python
# Contrast adjustment
alpha = 1.5  # contrast
beta = 0     # brightness
adjusted = cv2.convertScaleAbs(image, alpha=alpha, beta=beta)

# Display original and adjusted images
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1); plt.imshow(image); plt.title("Original Image"); plt.axis('off')
plt.subplot(1, 2, 2); plt.imshow(adjusted); plt.title("Contrast Adjusted"); plt.axis('off')
plt.show()
```

## Edge Enhancement
Enhance the edges in the image using the Canny edge detection algorithm.

```python
# Edge enhancement using Canny
gray = cv2.cvtColor(image, cv2.COLOR_RGB2GRAY)
edges = cv2.Canny(gray, 100, 200)

# Display original and edge-detected images
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1); plt.imshow(image); plt.title("Original Image"); plt.axis('off')
plt.subplot(1, 2, 2); plt.imshow(edges, cmap='gray'); plt.title("Canny Edges"); plt.axis('off')
plt.show()
```

## Color Space Conversion
Convert the image to the HSV color space and display the Hue, Saturation, and Value channels.

```python
# Convert to HSV and split channels
hsv = cv2.cvtColor(image, cv2.COLOR_RGB2HSV)
h, s, v = cv2.split(hsv)

# Display original and HSV channels
plt.figure(figsize=(12, 6))
plt.subplot(1, 4, 1); plt.imshow(image); plt.title("Original Image"); plt.axis('off')
plt.subplot(1, 4, 2); plt.imshow(h, cmap='gray'); plt.title("Hue"); plt.axis('off')
plt.subplot(1, 4, 3); plt.imshow(s, cmap='gray'); plt.title("Saturation"); plt.axis('off')
plt.subplot(1, 4, 4); plt.imshow(v, cmap='gray'); plt.title("Value"); plt.axis('off')
plt.show()
```

## Noise Reduction
Apply Gaussian and Median blurring to reduce noise in the image.

```python
# Apply Gaussian and Median blurring
gaussian = cv2.GaussianBlur(image, (5, 5), 0)
median = cv2.medianBlur(image, 5)

# Display original and blurred images
plt.figure(figsize=(12, 6))
plt.subplot(1, 3, 1); plt.imshow(image); plt.title("Original Image"); plt.axis('off')
plt.subplot(1, 3, 2); plt.imshow(gaussian); plt.title("Gaussian Blur"); plt.axis('off')
plt.subplot(1, 3, 3); plt.imshow(median); plt.title("Median Blur"); plt.axis('off')
plt.show()
```

## Binarization
Convert the image to a binary image using thresholding.

```python
# Convert to grayscale and apply thresholding
gray = cv2.cvtColor(image, cv2.COLOR_RGB2GRAY)
_, binary = cv2.threshold(gray, 128, 255, cv2.THRESH_BINARY)

# Display original and binarized images
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1); plt.imshow(image); plt.title("Original Image"); plt.axis('off')
plt.subplot(1, 2, 2); plt.imshow(binary, cmap='gray'); plt.title("Binarized Image"); plt.axis('off')
plt.show()
```

## Resizing and Normalization
Resize the image to 256x256 pixels and normalize the pixel values to the range [0, 1].

```python
# Resize and normalize the image
resized = cv2.resize(image, (256, 256))
normalized = resized / 255.0

# Display original and normalized images
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1); plt.imshow(image); plt.title("Original Image"); plt.axis('off')
plt.subplot(1, 2, 2); plt.imshow(normalized); plt.title("Normalized Image"); plt.axis('off')
plt.show()
```

## Sharpening
Sharpen the image using a kernel.

```python
# Apply sharpening kernel
kernel = np.array([[-1, -1, -1], [-1, 9, -1], [-1, -1, -1]])
sharpened = cv2.filter2D(image, -1, kernel)

# Display original and sharpened images
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1); plt.imshow(image); plt.title("Original Image"); plt.axis('off')
plt.subplot(1, 2, 2); plt.imshow(sharpened); plt.title("Sharpened Image"); plt.axis('off')
plt.show()
```

### Visualizing Intermediate Results
Below are the intermediate results for each preprocessing technique.

```python
# Display intermediate results for all techniques
techniques = {
    'Contrast Adjustment': adjusted,
    'Edge Enhancement': edges,
    'Color Space Conversion': hsv,
    'Noise Reduction (Gaussian)': gaussian,
    'Noise Reduction (Median)': median,
    'Binarization': binary,
    'Resizing & Normalization': normalized,
    'Sharpening': sharpened
}

for technique, result in techniques.items():
    plt.figure(figsize=(5, 5))
    plt.title(technique)
    if technique == 'Binarization':
        plt.imshow(result, cmap='gray')
    else:
        plt.imshow(result)
    plt.axis('off')
    plt.show()
```

### Summary of Preprocessing Techniques
This notebook demonstrated the following preprocessing techniques:
- **Contrast Adjustment**: Enhanced image contrast using scaling and offset.
- **Edge Enhancement**: Detected edges using the Canny algorithm.
- **Color Space Conversion**: Converted the image to HSV and displayed its channels.
- **Noise Reduction**: Reduced noise using Gaussian and Median blurring.
- **Binarization**: Converted the image to binary using thresholding.
- **Resizing and Normalization**: Resized the image and normalized pixel values.
- **Sharpening**: Enhanced image details using a sharpening kernel.
